### Feature Engineering

```python
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
```

```python
df['price_per_sqft'] = df['price'] / df['living_area_sqft']
```

```python
df['total_rooms'] = df['bedrooms'] + df['bathrooms']
df['bathroom_ratio'] = df['bathrooms'] / df['bedrooms']
```

```python
df['house_age'] = df['year'] - df['year_built']
df['years_since_renovation'] = df['year'] - df['year_renovated']
df.loc[df['years_since_renovation'] < 0, 'years_since_renovation'] = 0

```

```python
df['bedroom_category'] = pd.cut(df['bedrooms'], bins=[0,2,4,6,12], labels=['Small','Medium','Large','Huge'])
```

## Visualization

#### 1) Price vs Size Scatter 

```python
fig = px.scatter(df, x='living_area_sqft', y='price', size='bedrooms', color='condition_rating',
                 hover_data=['zipcode', 'grade_rating'], size_max=30,
                 title='Price vs Living Area: Size, Condition & Bedrooms')
fig.show()
```

#### 2) Monthly Price Pattern

```python
monthly_price = df.groupby(['year','month'])['price'].mean().unstack()

plt.figure(figsize=(10,6))
for year in monthly_price.index:
    plt.plot(monthly_price.columns, monthly_price.loc[year], label=year)

plt.title('Monthly Average Price Trend by Year')
plt.xlabel('Month')
plt.ylabel('Average Price')
plt.legend(title='Year')
plt.show()
```

#### 3) Distribution Histograms 

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
sns.histplot(data=df, x='price', kde=True, ax=axes[0,0]).set_title('Price Distribution')
sns.histplot(data=df, x='price_per_sqft', kde=True, ax=axes[0,1]).set_title('Price per Sqft Distribution')
sns.histplot(data=df, x='living_area_sqft', kde=True, ax=axes[1,0]).set_title('Living Area Distribution')
sns.histplot(data=df, x='lot_area_sqft', kde=True, ax=axes[1,1]).set_title('Lot Area Distribution')
plt.tight_layout()
plt.show()
```

#### 4) Correlation Heatmap 

```python
plt.figure(figsize=(12, 8))
corr_matrix = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr_matrix[['price']].sort_values('price', ascending=False), 
            annot=True, cmap='coolwarm', center=0, linewidths=2,
            cbar_kws={'label': 'Correlation with Price'})
plt.title('Feature Correlations with Price')
plt.tight_layout()
plt.show()
```

#### 5) Bar Charts by Condition & Grade

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
cond_means = df.groupby('condition_rating')['price'].mean()
ax1.bar(cond_means.index, cond_means.values, color='skyblue', edgecolor='navy')
ax1.set_title('Avg Price by Condition Rating')
ax1.set_ylabel('Average Price ($)')
ax1.tick_params(axis='x', rotation=0)

# Horizontal bar: Price by Grade
grade_means = df.groupby('grade_rating')['price'].mean()
ax2.barh(grade_means.index, grade_means.values, color='coral', edgecolor='darkred')
ax2.set_title('Avg Price by Grade Rating')
ax2.set_xlabel('Average Price ($)')
plt.tight_layout()
plt.show()
```

#### 6) Stacked Bar: Rooms Breakdown

```python
room_df = df[['bedrooms', 'bathrooms', 'total_rooms']].copy()
room_df['bedrooms'] = room_df['bedrooms'].astype(str)
room_df['price_cat'] = pd.cut(df['price'], bins=3, labels=['Low', 'Medium', 'High'])

pivot_table = room_df.pivot_table(index='price_cat', columns='bedrooms', values='total_rooms', aggfunc='mean').fillna(0)
pivot_table.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')
plt.title('Total Rooms Composition by Price Category & Bedrooms')
plt.ylabel('Average Total Rooms')
plt.legend(title='Bedrooms', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

#### 7) Time Trends Line Charts

```python
fig = px.line(df.sort_values('date'), x='date', y='price', 
              markers=True, title='Price Trends Over Time',
              labels={'price': 'Sale Price ($)', 'date': 'Sale Date'})
fig.add_scatter(x=df['date'], y=df['price_per_sqft'], mode='lines+markers', 
                name='Price per Sqft', yaxis='y2')
fig.update_layout(yaxis2=dict(overlaying='y', side='right', title='Price/Sqft'))
fig.show()
```

#### 8) Bedrooms Impact on Price & Bathroom Ratio vs Price

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.barplot(x='bedrooms', y='price', data=df, estimator='mean', ax=axes[0])
axes[0].set_title('Average Price by Number of Bedrooms')
axes[0].set_xlabel('Bedrooms')
axes[0].set_ylabel('Average Price')

sns.scatterplot(x='bathroom_ratio', y='price', data=df, ax=axes[1])
axes[1].set_title('Bathroom Ratio vs House Price')
axes[1].set_xlabel('Bathroom Ratio')
axes[1].set_ylabel('Price')

plt.tight_layout()
plt.show()
```

#### 9) Grade vs Condition Impact

```python
grade_condition = df.groupby(['grade_rating','condition_rating']).size().unstack(fill_value=0)

grade_condition.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title('House Count by Grade and Condition')
plt.xlabel('Grade Rating')
plt.ylabel('Number of Houses')
plt.legend(title='Condition Rating')
plt.show()
```

#### 10) Zipcode vs Price

```python
zip_price = df.groupby('zipcode')['price'].mean().reset_index()

fig = px.bar(
    zip_price,
    x='zipcode',
    y='price',
    title='Average House Price by Zipcode',
    labels={'price':'Average Price'}
)
fig.show()
```

#### 11)  House Age Impact

```python
age_group = pd.cut(
    df['house_age'],
    bins=[0,50,100,150],
    labels=['0–50 yrs','50–100 yrs','100–150 yrs']
)

age_price = df.groupby(age_group)['price'].mean()

plt.figure(figsize=(8,5))
plt.barh(age_price.index, age_price.values)
plt.title('Average Price by House Age Group')
plt.xlabel('Average Price')
plt.ylabel('House Age Group')
plt.show()
```

## Summary

## Data Analysis Key Findings
* **Price vs Size Scatter** shows a strong positive relationship between living area and house price, confirming that larger houses generally have higher market value.
* **Monthly Price Pattern** highlights seasonal variations in housing prices, indicating that demand and pricing fluctuate across different months.
* **Distribution Histograms** reveal that house prices are right-skewed, with most properties concentrated in mid-price ranges and fewer luxury homes at higher prices.
* **Correlation Heatmap** identifies living area, grade rating, and above-ground square footage as the strongest predictors of price, while age-related and time-based features show weak influence.
* **Bar Charts by Condition & Grade** demonstrate that houses with better condition and higher grade ratings consistently achieve higher average prices.
* **Stacked Bar: Rooms Breakdown** compares room composition across properties and shows that houses with more usable rooms are generally associated with higher price categories.
* **Time Trends Line Charts** display overall price movement across years, helping to understand market stability and long-term trends.
* **Bedrooms Impact on Price & Bathroom Ratio vs Price** indicate that bathrooms and layout efficiency have a stronger impact on price than bedroom count alone.
* **Grade vs Condition Impact** highlights that construction quality (grade) combined with property condition significantly increases house value.
* **Zipcode vs Price** emphasizes the role of location, showing noticeable price differences across zipcodes even for similar property features.
* **House Age Impact** shows a slight negative relationship between house age and price, meaning newer or recently renovated houses tend to be priced higher.
* **Bathroom Ratio vs Price** confirms that a higher bathroom-to-bedroom ratio improves usability and contributes positively to house pricing.
